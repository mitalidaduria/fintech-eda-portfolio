```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../data/PS_20174392719_1491204439457_log.csv', nrows=500000)
```

```python
# 1. Basic Profiling
print("Dataset Shape:", df.shape)
print("\nMissing Values:\n", df.isnull().sum())
```

```python
# 2. Fraud Imbalance Check
fraud_counts = df['isFraud'].value_counts(normalize=True) * 100
print(f"\nFraudulent Transactions: {fraud_counts[1]:.4f}%")
print(f"Legitimate Transactions: {fraud_counts[0]:.4f}%")

# 3. Transaction Type Analysis
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='type', hue='isFraud')
plt.title('Fraud Distribution by Transaction Type')
plt.yscale('log') # Log scale because fraud is rare
plt.show()
```
